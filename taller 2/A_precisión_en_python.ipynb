```python

import math

def suma_seno(x, tol=1e-8, max_iter=200):
    # La serie de Taylor de sin(x) comienza con el término x.
    # 'term' guarda el término actual y 'total' acumula la aproximación.
    term = x
    total = x

    # Se generan nuevos términos hasta alcanzar la tolerancia
    # o llegar al número máximo de iteraciones.
    for i in range(max_iter):
        n = i + 1

        # Relación recursiva entre términos consecutivos:
        # evita calcular potencias y factoriales desde cero.
        term = -term * x**2 / ((2*n)*(2*n + 1))
        total += term

        # Se detiene cuando el término añadido es suficientemente
        # pequeño con respecto a la escala de la suma acumulada.
        if abs(term) < tol * max(1.0, abs(total)):
            break

    # Se devuelve la aproximación y el número de iteraciones utilizadas.
    return total, i + 1
```

```python

def suma_seno_reducida(x, tol=1e-8, max_iter=200):
    # El seno tiene periodo 2*pi, por lo que sin(x) = sin(x_reducido).
    # math.remainder reduce el argumento y ayuda a mejorar la convergencia.
    x_reducido = math.remainder(x, 2 * math.pi)

    # Se reutiliza la función principal para evitar duplicar el algoritmo.
    return suma_seno(x_reducido, tol, max_iter)
```

```python

import math  # Importa el módulo matemático para utilizar funciones como seno y valor absoluto.

# --- Verificación Parte A: suma_seno_reducida ---
# Se comprueba que la función suma_seno_reducida calcule correctamente el seno
# para diferentes valores de x, comparando sus resultados con math.sin().

casos_a = [
    # Cada tupla contiene: (valor de x, resultado esperado, tolerancia permitida).
    (0.5, math.sin(0.5), 1e-8),       # Caso con un argumento pequeño.
    (100.0, math.sin(100.0), 1e-6),   # Caso con un argumento positivo grande.
    (-237.0, math.sin(-237.0), 1e-6), # Caso con un argumento negativo grande.
]

# Se recorren los casos de prueba para verificar uno por uno los resultados.
for x, esperado, tol in casos_a:

    # Se ejecuta la función y se obtienen la aproximación del seno
    # y el número de iteraciones realizadas.
    obtenido, _ = suma_seno_reducida(x)

    # Se calcula el error absoluto entre la aproximación y el valor de referencia.
    error = abs(obtenido - esperado)

    # Se comprueba si el error es menor que la tolerancia establecida.
    # Si se cumple, la prueba se considera aprobada.
    ok = error < tol

    # Se asigna PASS cuando la prueba cumple la tolerancia y FAIL en caso contrario.
    estado = "PASS" if ok else "FAIL"

    # Se imprimen el estado, el argumento evaluado, la aproximación,
    # el valor de referencia y el error absoluto para analizar el resultado.
    print(
        f"{estado}  x={x:>8.1f}  "
        f"obtenido={obtenido: .10f}  "
        f"esperado={esperado: .10f}  "
        f"error={error:.3e}"
    )
```

la función suma_seno_reducida(x) supera las pruebas de verificación realizadas. Los resultados son consistentes con una implementación correcta de la serie de Taylor con reducción del argumento.

**2. Genere la tabla de convergencia (como en la sección 7) para al menos 6 valores de $x$ entre $0$ y $4\pi$, incluyendo el número de iteraciones, la suma y el error relativo frente a `math.sin`.**

```python

# Se define una lista de valores de prueba distribuidos entre 0 y 4*pi.
# Estos valores permiten analizar el comportamiento de la serie de Taylor
# para diferentes argumentos, incluyendo múltiplos de pi.
valores_x = [
    0,                  # El seno de cero es exactamente cero.
    math.pi / 4,        # Primer ángulo de prueba: pi/4.
    math.pi / 2,        # Ángulo en el que el seno vale uno.
    math.pi,            # El seno de pi es cero.
    2 * math.pi,        # El seno de 2*pi es cero.
    3 * math.pi,        # El seno de 3*pi es cero.
    4 * math.pi         # El seno de 4*pi es cero.
]

# Se imprime el encabezado de la tabla de resultados.
# Los números después de ":" indican el ancho de cada columna.
# El símbolo ">" alinea el texto hacia la derecha.
print(f"{'x':>12} {'Iteraciones':>12} {'Suma':>20} {'Error':>20}")

# Se recorre la lista para evaluar la función en cada valor de x.
for x in valores_x:

    # Se ejecuta la función suma_seno(x), que calcula una aproximación
    # del seno mediante la serie de Taylor.
    # La función devuelve la suma aproximada y el número de iteraciones.
    suma, iteraciones = suma_seno(x)

    # Se calcula el valor de referencia usando la función seno
    # de la biblioteca matemática de Python.
    exacto = math.sin(x)

    # Se selecciona el tipo de error según el valor de referencia.
    # Si el seno es suficientemente diferente de cero, se calcula
    # el error relativo para comparar la diferencia con el valor exacto.
    if abs(exacto) > 1e-15:
        error = abs((suma - exacto) / exacto)

    # Cuando el seno de referencia es cero o está muy cerca de cero,
    # se utiliza el error absoluto para evitar dividir por un número
    # demasiado pequeño, lo que produciría un error relativo poco útil.
    else:
        error = abs(suma - exacto)

    # Se muestran los resultados de cada prueba en una fila.
    # x se presenta con 6 decimales y la suma y el error en notación científica.
    # La cantidad de iteraciones se presenta como un número entero.
    print(f"{x:12.6f} {iteraciones:12d} {suma:20.12e} {error:20.12e}")
```

Los resultados muestran que la función `suma_seno(x)` aproxima correctamente el seno para los valores evaluados. A medida que aumenta \(x\), generalmente se necesitan más iteraciones para calcular la suma. El error máximo observado fue aproximadamente $2.51\times10^{-10}$, lo que indica una buena precisión. Las pequeñas diferencias en los múltiplos de $\pi$ se deben a los errores de redondeo numérico.

```python

# Se establece la tolerancia exigida para evaluar la precisión.
tol = 1e-8

# Se muestra la tolerancia en notación científica.
print(f"Tolerancia exigida: {tol:.1e}\n")

# Se recorren los valores de x utilizados en el Punto 2.
for x in valores_x:

    # Se calcula la aproximación del seno y el número de iteraciones.
    suma, iteraciones = suma_seno(x)

    # Se obtiene el valor de referencia mediante la función matemática de Python.
    referencia = math.sin(x)

    # Si el valor de referencia no está cerca de cero,
    # se calcula el error relativo.
    if abs(referencia) > 1e-15:
        error = abs((suma - referencia) / referencia)
        tipo_error = "relativo"

    # Si el valor de referencia está cerca de cero,
    # se utiliza el error absoluto para evitar dividir por un número muy pequeño.
    else:
        error = abs(suma - referencia)
        tipo_error = "absoluto"

    # Se compara el error calculado con la tolerancia exigida.
    # Se cumple si el error es menor que la tolerancia.
    estado = "CUMPLE" if error < tol else "NO CUMPLE"

    # Se muestran el argumento, las iteraciones, el tipo de error,
    # su valor numérico y el resultado de la comparación.
    print(
        f"x={x:10.6f} | iteraciones={iteraciones:3d} | "
        f"error {tipo_error}={error:.3e} | {estado}"
    )
```

No se puede afirmar que el error relativo sea menor que $10^{-8}$ en todos los casos. Para $x=n\pi$, el valor teórico de $\sin(x)$ es cero, por lo que el error relativo no está definido. En estos casos se utiliza el error absoluto, cuyo valor sí permite evaluar la precisión de la aproximación. Para los demás valores evaluados, el error relativo cumple la tolerancia establecida. Por tanto, la función ofrece buenas aproximaciones, pero es necesario distinguir entre error relativo y absoluto.

**5. Compare su implementación recursiva con una versión "ingenua" que calcule cada término desde cero usando `math.factorial` y `x**potencia` — ¿en qué valores de $x$ empieza a notarse la diferencia de desempeño o de precisión?**

```python
# --- Versión ingenua de la serie de Taylor del seno ---

def suma_seno_ingenua(x, tol=1e-8, max_iter=200):

    # Se inicializa la suma total en cero para acumular los términos.
    total = 0

    # Se recorren los términos de la serie hasta alcanzar la tolerancia
    # o llegar al número máximo de iteraciones permitido.
    for n in range(max_iter):

        # Se calcula el exponente impar correspondiente al término actual:
        # 1, 3, 5, 7, ...
        potencia = 2*n + 1

        # Se calcula el término completo desde cero:
        # (-1)^n determina el signo alternante,
        # x^potencia calcula la potencia de x,
        # y factorial(potencia) calcula el factorial del exponente.
        term = (-1)**n * x**potencia / math.factorial(potencia)

        # Se suma el término actual al resultado acumulado.
        total = total + term

        # Se verifica si el término es suficientemente pequeño
        # respecto a la tolerancia y al tamaño de la suma acumulada.
        # Si se cumple la condición, se termina el ciclo.
        if abs(term) < tol * max(1.0, abs(total)):
            break

    # Se devuelve la suma aproximada y el número de términos utilizados.
    return total, n + 1
```

```python

# --- Comparación entre versión recursiva e ingenua ---

# Se seleccionan valores pequeños y grandes para estudiar
# el comportamiento de ambos métodos.
valores_x = [1, 5, 10, 20, 50, 100]

# Se imprimen los encabezados de la tabla comparativa.
# El símbolo > alinea el contenido a la derecha.
print(f"{'x':>6} {'Recursiva':>15} {'Ingenua':>15} "
      f"{'Error recursiva':>18} {'Error ingenua':>18}")

# Se imprime una línea para separar los encabezados de los resultados.
print("-" * 80)

# Se repite la comparación para cada valor de x.
for x in valores_x:

    # Se calcula la aproximación con la versión recursiva
    # y se guarda el número de iteraciones realizadas.
    recursiva, iter_rec = suma_seno(x)

    # Se calcula la aproximación con la versión ingenua
    # y se guarda el número de términos utilizados.
    ingenua, iter_ing = suma_seno_ingenua(x)

    # Se obtiene el valor de referencia mediante math.sin().
    exacto = math.sin(x)

    # Se calcula el error absoluto de la versión recursiva.
    error_rec = abs(recursiva - exacto)

    # Se calcula el error absoluto de la versión ingenua.
    error_ing = abs(ingenua - exacto)

    # Se muestran las aproximaciones y sus errores en notación científica.
    print(f"{x:6.1f} {recursiva:15.8f} {ingenua:15.8f} "
          f"{error_rec:18.3e} {error_ing:18.3e}")
```


Para $x=100$, la versión recursiva tarda aproximadamente $3.80\times10^{-5},\text{s}$, mientras que la versión ingenua tarda $3.60\times10^{-4},\text{s}$. Por tanto, la versión recursiva es aproximadamente $9.5$ veces más rápida. Esto se debe a que reutiliza el término anterior, mientras que la ingenua calcula las potencias y los factoriales desde cero.




```python

# --- Comparación de desempeño ---

# Importar el módulo para medir tiempos de ejecución.
import time

# Seleccionar valores pequeños y grandes para comparar ambos métodos.
valores_x = [1, 5, 10, 20, 50, 100]

# Definir cuántas veces se ejecutará cada función para obtener
# un tiempo promedio más estable.
repeticiones = 100

# Mostrar los encabezados de la tabla.
print(f"{'x':>6} {'Tiempo recursiva (s)':>22} {'Tiempo ingenua (s)':>22}")
print("-" * 55)

# Recorrer cada valor de x.
for x in valores_x:

    # Registrar el instante inicial para medir la función recursiva.
    inicio = time.perf_counter()

    # Ejecutar la función recursiva varias veces.
    for _ in range(repeticiones):
        suma_seno(x)

    # Calcular el tiempo promedio de una ejecución recursiva.
    tiempo_rec = (time.perf_counter() - inicio) / repeticiones

    # Registrar el instante inicial para medir la función ingenua.
    inicio = time.perf_counter()

    # Ejecutar la función ingenua varias veces.
    for _ in range(repeticiones):
        suma_seno_ingenua(x)

    # Calcular el tiempo promedio de una ejecución ingenua.
    tiempo_ing = (time.perf_counter() - inicio) / repeticiones

    # Mostrar x y los tiempos promedio en segundos.
    print(f"{x:6.1f} {tiempo_rec:22.10e} {tiempo_ing:22.10e}")
```

La función recursiva fue más rápida en todos los valores evaluados. Para $x=1$, los tiempos son similares, pero la diferencia aumenta al crecer $x$. Para $x=100$, la función ingenua tarda aproximadamente $9.5$ veces más que la recursiva. Esto ocurre porque la recursiva reutiliza el término anterior, mientras que la ingenua calcula potencias y factoriales desde cero. Por tanto, la implementación recursiva es más eficiente, aunque ambas pueden perder precisión para valores grandes de $x$ sin reducción de periodicidad.

En cuanto a la precisión, para $x = 20$ la versión ingenua resulta ligeramente más precisa que la recursiva ($2.05 \times 10^{-9}$ frente a $4.88 \times 10^{-9}$). Esto se debe a que la versión recursiva acumula el error de redondeo (floating-point round-off error) en cada iteración al multiplicar por el término anterior, mientras que la versión ingenua evalúa cada término de manera independiente.
